7. Study Hours vs Marks

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Study_Hours', y='Marks')
plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

8. Attendance vs Marks

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Attendance_Percent', y='Marks')
plt.title('Attendance vs Marks')
plt.xlabel('Attendance Percentage')
plt.ylabel('Marks')
plt.show()
```

9. Screen Time vs Marks

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Screen_Time_Hours', y='Marks')
plt.title('Screen Time vs Marks')
plt.xlabel('Screen Time Hours')
plt.ylabel('Marks')
plt.show()
```

10. Sleep Hours vs Marks

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Sleep_Hours', y='Marks')
plt.title('Sleep Hours vs Marks')
plt.xlabel('Sleep Hours')
plt.ylabel('Marks')
plt.show()
```

11. Marks Histogram

```python
plt.figure(figsize=(8,5))
plt.hist(df['Marks'], bins=10)
plt.title('Distribution of Marks')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.show()
```

12. Correlation Heatmap

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

13. Marks Box Plot

```python
plt.figure(figsize=(6,5))
sns.boxplot(y=df['Marks'])
plt.title('Marks Distribution')
plt.ylabel('Marks')
plt.show()
```
